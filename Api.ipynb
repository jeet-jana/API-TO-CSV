```python
import pandas as pd
import requests
```

```python
response = requests.get("https://goodrecmovies.com/api/list?kind=movie&pageSize=10")
```

```python
df = pd.DataFrame(response.json()['rows'])[['startYear','adult','avgRating','votes','rank']]
```

```python
df = pd.DataFrame()
```

```python
for i in range(0,349):
    response = requests.get(f"https://goodrecmovies.com/api/list?kind=movie&pageSize=10&page={i}")
    temp_df = pd.DataFrame(response.json()['rows'])[['startYear','adult','avgRating','votes','rank']]
    df = pd.concat([df,temp_df],ignore_index=True)
print("Success full")
```
